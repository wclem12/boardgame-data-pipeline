# 01 – Explore the Board Game Data

**Purpose:** First look at the BoardGameGeek data before building anything. Check the size of each file, the columns, how the list columns are stored, and how well the two files match.

**Data:** `ratings.csv` and `details.csv` from [Board Games on Kaggle](https://www.kaggle.com/datasets/joebeachcapital/board-games) (originally BoardGameGeek via TidyTuesday, January 2022). The files are uploaded to Colab each session and are not stored in this repo.

**Findings**
- `ratings.csv`: 21,831 games, 10 columns. `details.csv`: 21,631 games, 23 columns.
- 200 rated games have no matching details. The join type will decide whether they're kept.
- There is no complexity column.
- `details.csv` includes publisher, plus popularity columns: `owned`, `trading`, `wanting`, `wishing`.
- List columns (mechanics, categories, designers) are stored as text that looks like a Python list. Some values contain commas inside the name (e.g. `'Deck, Bag, and Pool Building'`), so splitting on commas would break them.
- `name` vs. `primary` and `year` vs. `yearpublished` look like duplicates. Still to check whether they ever disagree.
